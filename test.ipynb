```python
import requests
url = "https://api.alertaq.com/api/v4/public/locations"
property_to_find = 'BGO Pen Centre'
response = requests.get(url, headers={"Token": token})
response.raise_for_status()  # Raise an exception for HTTP errors

locations = response.json()
properties = locations.get('dataModel', [])
if not properties:
    print("No properties found in the response.")

matching_properties = [p for p in properties if p.get('name') == 'BGO Pen Center']
```
